# causal information in the baseline execution

eight nodes, thirty independent seeds.

```python
from pathlib import Path

import pandas as pd

from causality_bench.provenance import read_csv

pd.set_option("display.width", 200)

# results/ sits beside notebooks/ in the repo root
RESULTS = Path.cwd().parent / "results"


def raw(name):
    return read_csv(RESULTS / "raw" / f"{name}.csv")


def processed(name):
    return read_csv(RESULTS / "processed" / f"{name}.csv")
```

the baseline metrics, mean and standard deviation across the seeds.

```python
metrics = [
    "causal_density",
    "concurrency_ratio",
    "lamport_false_ordering_rate",
    "lamport_concurrency_recall",
    "lamport_order_precision",
]
summary = processed("baseline_summary")
summary[["nodes", "seeds"] + [f"{m}_mean" for m in metrics] + [f"{m}_std" for m in metrics]].T
```

the tied pairs are the only concurrency a scalar clock can still express; everything else in the concurrent set is given an order the execution does not contain.

```python
baseline = raw("baseline")
baseline["misrepresented_share"] = baseline.lamport_false_orderings / baseline.total_pairs

baseline[["events", "concurrent_pairs", "lamport_tied_pairs", "lamport_false_orderings",
          "misrepresented_share"]].describe().T
```
